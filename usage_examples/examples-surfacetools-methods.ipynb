### 1.7 `get_boundary_edges`

Extract only the boundary edges from a triangular mesh.

```python
########################### Testing get_boundary_edges ###########################
import clabtoolkit.surfacetools as cltsurf
import pyvista as pv

############## Creating a Surface with edges ###################################
# Create outer boundary points (square)
outer_boundary = np.array([
    [-1, -1, 0],
    [1, -1, 0],
    [1, 1, 0],
    [-1, 1, 0]
])

# Create hole boundary points (circle)
theta = np.linspace(0, 2*np.pi, 30, endpoint=False)
hole_boundary = np.column_stack([
    0.3 * np.cos(theta),
    0.3 * np.sin(theta),
    np.zeros_like(theta)
])

# Combine all boundary points
all_points = np.vstack([outer_boundary, hole_boundary])

# Create point cloud
point_cloud = pv.PolyData(all_points)

# Perform 2D Delaunay triangulation
mesh = point_cloud.delaunay_2d()

print("\nTest 1: Loading surface and getting edges...")
surf_obj = cltsurf.Surface(mesh)

b_edges = surf_obj.get_boundary_edges()
print(f"Number of boundary edges in the surface: {len(b_edges)}")
print(f"First 3 edges: {b_edges[:3]}")
```

### 1.10 `get_info`

Display and return comprehensive information about the Surface object.

```python
########################### Testing get_info ###########################
import clabtoolkit.surfacetools as cltsurf
import os
import clabtoolkit.misctools as cltmisc

############## Loading from a surface file ###################################
# Sample data of clabtoolkit (the surfaces and the maps are downloaded from Zenodo the first time)
surf_dir = cltmisc.get_sample_data_dir("surfaces")
annot_dir = cltmisc.get_sample_data_dir("annotations")
maps_dir = cltmisc.get_sample_data_dir("corticalmaps")
prefix = "sub-colin27_ses-01_hemi-L_space-tallin"
surf_file = os.path.join(surf_dir, f"{prefix}_pial.surf.gii")
annot_file = os.path.join(annot_dir, f"{prefix}_atlas-Desikan2006_dseg.label.gii")
thickness_file = os.path.join(maps_dir, f"{prefix}_thickness.shape.gii")

surf_obj = cltsurf.Surface(surf_file, name="colin27_lh_pial")

print("\nTest 1: Information about a freshly loaded surface...")
# The name is optional. If it is not supplied the surface will be named "unknown"
surf_obj.get_info()
print('')

print("\nTest 2: Information after adding a parcellation, a map and the normals...")
surf_obj.load_annotation(annot_file, parc_name="Desikan")
surf_obj.load_scalar_maps(thickness_file)
surf_obj.compute_normals()
surf_obj.get_info()
print('')

print("\nTest 3: Getting the information as a dictionary (verbose=False)...")
info = surf_obj.get_info(verbose=False)
print(f"Surface name       : {info['name']}")
print(f"Hemisphere         : {info['hemi']}")
print(f"Number of vertices : {info['n_vertices']}")
print(f"Number of faces    : {info['n_faces']}")
print(f"Global color       : {info['global_color']}")
print(f"Global opacity     : {info['global_opacity']}")

# Vertex-wise maps with and without colortable
print("Vertex-wise maps:")
for overlay in info['overlays']:
    if overlay['has_colortable']:
        print(f"  - {overlay['name']} ({overlay['type']}): colortable with {overlay['n_regions']} region(s)")
    else:
        print(f"  - {overlay['name']} ({overlay['type']}): no colortable")
```

### 1.23 `apply_affine`

Apply an affine transformation to the surface mesh.

```python
######################### Testing apply_affine ###########################
import clabtoolkit.surfacetools as cltsurf
import os
import clabtoolkit.misctools as cltmisc
import numpy as np


# Sample data of clabtoolkit (the surfaces and the maps are downloaded from Zenodo the first time)
surf_dir = cltmisc.get_sample_data_dir("surfaces")
prefix = "sub-colin27_ses-01_hemi-L_space-tallin"
surf_file = os.path.join(surf_dir, f"{prefix}_pial.surf.gii")

surf_obj = cltsurf.Surface(surf_file)

# Test 1: Applying an affine transformation to the surface
print("Test 1: Applying an affine transformation to the surface and modifying the object...")
# Define an affine transformation matrix (4x4)
affine_matrix = np.array([[-1, 0, 0, 133.0],  # Translation in x
                        [0, 1, 0, -110.0],  # Translation in y
                        [0, 0, 1, -128.0],  # Translation in z
                        [0, 0, 0, 1]])      # Homogeneous

print("Original coordinates limits:")
# Max of the original vertices before transformation
print("Max of each coordinate of the original vertices before transformation:")
print(f"Max X: {surf_obj.mesh.points[:,0].max()}")
print(f"Max Y: {surf_obj.mesh.points[:,1].max()}")
print(f"Max Z: {surf_obj.mesh.points[:,2].max()}")
print("Min of each coordinate of the original vertices before transformation:")
print(f"Min X: {surf_obj.mesh.points[:,0].min()}")
print(f"Min Y: {surf_obj.mesh.points[:,1].min()}")
print(f"Min Z: {surf_obj.mesh.points[:,2].min()}")

print( " ")

surf_obj.apply_affine(affine_matrix)
print("Applying affine transformation matrix:")
print("Max of each coordinate of the transformed vertices:")
print(f"Max X: {surf_obj.mesh.points[:,0].max()}")
print(f"Max Y: {surf_obj.mesh.points[:,1].max()}")
print(f"Max Z: {surf_obj.mesh.points[:,2].max()}")
print("Min of each coordinate of the transformed vertices:")
print(f"Min X: {surf_obj.mesh.points[:,0].min()}")
print(f"Min Y: {surf_obj.mesh.points[:,1].min()}")
print(f"Min Z: {surf_obj.mesh.points[:,2].min()}")

# Test 2: Applying the inverse of an affine transformation to the surface
print("Test 2: Applying the inverse of an affine transformation to the surface and modifying the object...")
# Apply the inverse of the affine transformation to the surface
print( " ")

surf_obj.apply_affine(affine_matrix, inverse=True)
print("Inverse affine transformation applied successfully.")
print("Max of each coordinate of the transformed vertices after applying the inverse:")
print(f"Max X: {surf_obj.mesh.points[:,0].max()}")
print(f"Max Y: {surf_obj.mesh.points[:,1].max()}")
print(f"Max Z: {surf_obj.mesh.points[:,2].max()}")
print("Min of each coordinate of the transformed vertices after applying the inverse:")
print(f"Min X: {surf_obj.mesh.points[:,0].min()}")
print(f"Min Y: {surf_obj.mesh.points[:,1].min()}")
print(f"Min Z: {surf_obj.mesh.points[:,2].min()}")
```
