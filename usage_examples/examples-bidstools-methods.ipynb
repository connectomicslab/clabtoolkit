### 1.1 `str2entity`
<a id="bidstools-str2entity"></a>

Convert a BIDS-formatted string into an entity dictionary.

```python
######################## Testing str2entity ###################################

import clabtoolkit.bidstools as cltbids

print('Test 1: Parse a standard T1w filename with multiple entities:')
filename = 'sub-01_ses-M00_acq-3T_dir-AP_run-01_T1w.nii.gz'
entity = cltbids.str2entity(filename)
print(f"Filename: {filename}")
print(entity)
print('')

print('Test 2: Parse a bold functional filename:')
filename = 'sub-02_ses-baseline_task-rest_run-01_bold.nii.gz'
entity = cltbids.str2entity(filename)
print(f"Filename: {filename}")
print(entity)
print('')

print('Test 3: Parse a DWI filename without a run entity:')
filename = 'sub-03_ses-M06_acq-b1000_dir-PA_dwi.nii.gz'
entity = cltbids.str2entity(filename)
print(f"Filename: {filename}")
print(entity)
print('')

print('Test 4: Parse a derivatives filename with a desc entity:')
filename = 'sub-01_ses-M00_desc-preproc_T1w.nii.gz'
entity = cltbids.str2entity(filename)
print(f"Filename: {filename}")
print(entity)
print('')

print('Test 5: Parse a filename without an extension (bare suffix only):')
filename = 'sub-01_ses-M00_T1w'
entity = cltbids.str2entity(filename)
print(f"Filename: {filename}")
print(entity)
print('')
```

### 1.2 `entity2str`
<a id="bidstools-entity2str"></a>

Convert an entity dictionary back to a BIDS-formatted string.

```python
######################## Testing entity2str ###################################

import clabtoolkit.bidstools as cltbids

print('Test 1: Reconstruct a T1w filename from a full entity dictionary:')
entity = {'sub': '01', 'ses': 'M00', 'acq': '3T', 'dir': 'AP', 'run': '01',
        'suffix': 'T1w', 'extension': 'nii.gz'}
result = cltbids.entity2str(entity)
print(result)
print('')

print('Test 2: Reconstruct a bold filename from a dictionary without extension:')
entity = {'sub': '02', 'ses': 'baseline', 'task': 'rest', 'run': '01', 'suffix': 'bold'}
result = cltbids.entity2str(entity)
print(result)
print('')

print('Test 3: Verify round-trip consistency (str2entity -> entity2str):')
original = 'sub-03_ses-M06_acq-b1000_dir-PA_dwi.nii.gz'
entity = cltbids.str2entity(original)
reconstructed = cltbids.entity2str(entity)
print(f'  Original:      {original}')
print(f'  Reconstructed: {reconstructed}')
print(f'  Match: {original == reconstructed}')
print('')
```

### 1.3 `delete_entity`
<a id="bidstools-delete_entity"></a>

Remove specified entities from a dictionary or string.

```python
######################## Testing delete_entity ###################################

import clabtoolkit.bidstools as cltbids

filename = 'sub-01_ses-M00_acq-3T_dir-AP_run-01_T1w.nii.gz'

print('Test 1: Remove a single entity (acq) by key name from a string:')
result = cltbids.delete_entity(filename, 'acq')
print(f"Filename: {filename}")
print(result)
print('')

print('Test 2: Remove multiple entities (acq and dir) from a string:')
result = cltbids.delete_entity(filename, ['acq', 'dir'])
print(f"Filename: {filename}")
print(result)
print('')

print('Test 3: Remove entity only if it matches a specific value (dict input):')
result = cltbids.delete_entity(filename, {'acq': ['3T']})
print(f"Filename: {filename}")
print(result)
print('')

print('Test 4: Remove entity only if value matches — no match, nothing removed:')
result = cltbids.delete_entity(filename, {'acq': ['7T']})
print(f"Filename: {filename}")
print(result)
print('')

print('Test 5: Remove entity from a dictionary (dict input/output):')
entity = cltbids.str2entity(filename)
result = cltbids.delete_entity(entity, 'run')
print(result)
print('')
```

### 1.4 `replace_entity_value`
<a id="bidstools-replace_entity_value"></a>

Replace entity values in a dictionary or string.

```python
######################## Testing replace_entity_value ###################################

import clabtoolkit.bidstools as cltbids

filename = 'sub-01_ses-M00_acq-3T_dir-AP_run-01_T1w.nii.gz'

print('Test 1: Replace the acq value from 3T to 7T in a string:')
result = cltbids.replace_entity_value(filename, {'acq': '7T'})
print(f"Filename: {filename}")
print(result)
print('')

print('Test 2: Replace multiple entity values at once:')
result = cltbids.replace_entity_value(filename, {'acq': '7T', 'dir': 'PA', 'run': '02'})
print(f"Filename: {filename}")
print(result)
print('')

print('Test 3: Replace value using a BIDS string as the replacement dictionary:')
result = cltbids.replace_entity_value(filename, 'acq-7T_dir-PA')
print(f"Filename: {filename}")
print(result)
print('')

print('Test 4: Attempt to replace a non-existent key with verbose=True:')
result = cltbids.replace_entity_value(filename, {'task': 'rest'}, verbose=True)
print(f"Filename: {filename}")
print(result)
print('')

print('Test 5: Replace entity value in a dictionary (dict input/output):')
entity = cltbids.str2entity(filename)
result = cltbids.replace_entity_value(entity, {'ses': 'M24'})
print(f"Filename: {filename}")
print(result)
print('')
```

### 1.6 `insert_entity`

### 2.5 `is_bids_filename`
<a id="bidstools-is_bids_filename"></a>

Validate whether a filename follows the BIDS naming convention.

```python
######################## Testing is_bids_filename ###################################

import clabtoolkit.bidstools as cltbids

print('Test 1: Valid BIDS T1w filename:')
print(cltbids.is_bids_filename('sub-01_ses-M00_acq-3T_T1w.nii.gz'))   # True
print('')

print('Test 2: Valid BIDS bold filename:')
print(cltbids.is_bids_filename('sub-02_task-rest_run-01_bold.nii.gz'))  # True
print('')

print('Test 3: Valid BIDS filename with full path (path is stripped):')
print(cltbids.is_bids_filename('/data/sub-01/anat/sub-01_T1w.nii.gz'))  # True
print('')

print('Test 4: Invalid — no entity-label pairs, plain name:')
print(cltbids.is_bids_filename('my_custom_file.nii.gz'))                # False
print('')

print('Test 5: Invalid — suffix only, no entities:')
print(cltbids.is_bids_filename('T1w.nii.gz'))                           # False
print('')

print('Test 6: Invalid — key contains special characters:')
print(cltbids.is_bids_filename('sub-01_sesM00_T1w.nii.gz'))            # False
print('')

print('Test 7: Invalid entity:')
print(cltbids.is_bids_filename('sub-01_sesssion-M00_T1w.nii.gz', extensive=True))            # False
print('')   

print('Test 8: Invalid suffix:')
print(cltbids.is_bids_filename('sub-01_ses-M00_T1w1.nii.gz', extensive=True))            # False
print('')
```
