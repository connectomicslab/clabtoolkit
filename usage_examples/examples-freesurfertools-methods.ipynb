### 1.6 `fill_parcellation`
<a id="annotparc-fill_parcellation"></a>

Fill unlabeled vertices by iterative neighbor propagation.

```python
######################## Testing fill_parcellation ###################################
# This example always uses simulated data, so that the unlabeled vertices are known.
import os
import tempfile

import numpy as np

import clabtoolkit.freesurfertools as cltfree
import clabtoolkit.surfacetools as cltsurf
import clabtoolkit.visualizationtools as cltvis

############## Simulating the example data ###################################
# 1. Simulate a surface and save it to a temporary file
surface = cltsurf.Surface.simulate_surface()
surf_file = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.surface', delete=False).name
surface.save_surface(filename=surf_file, overwrite=True)

# 2. Simulate an annotation on the surface
annot_parc = cltfree.AnnotParcellation.simulate_annotation(surface=surface, n_regions=5)

# 3. Simulate a cortex label by excluding the last region
codes = np.asarray(annot_parc.codes).ravel()
last_code = np.unique(codes)[-1]              # code of the "last" region present on the surface
labeled_indices = np.where(codes != last_code)[0]
labeled_vertices = surface.mesh.points[labeled_indices]
label_data = np.concatenate([labeled_indices[:, None], labeled_vertices], axis=1)

# 4. Write the cortex label as a FreeSurfer .label file: [index, x, y, z, value]
label_file = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.label', delete=False).name
with open(label_file, 'w') as f:
    f.write('#!ascii label\n')
    f.write(f'{len(label_data)}\n')
    for idx, x, y, z in label_data:
        f.write(f'{int(idx)} {x:.6f} {y:.6f} {z:.6f} 0.500000\n')

# 5. Set some vertices of the cortex label as unlabeled
n_unlabeled = 20
unlabeled_indices = labeled_indices[:n_unlabeled]
annot_parc.codes[unlabeled_indices] = 0
print(f'Set {n_unlabeled} vertices as unlabeled in the annotation.')

# 6. Save the simulated annotation to a temporary file
annot_file = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.annot', delete=False).name
annot_parc.save_annotation(annot_file)
annot_id = 'Simulated'

out_annot = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.annot', delete=False).name

# Test 1: Fill the unlabeled vertices inside the cortex label
print('Test 1: Filling unlabeled vertices of the annotation inside the cortex label...')
annot_parc = cltfree.AnnotParcellation(parc_file=annot_file)
corrected_path, vert_lab, reg_ctable, reg_names = annot_parc.fill_parcellation(
    label_file=label_file,
    surf_file=surf_file,
    corr_annot=out_annot,
)
print(f'  Corrected annotation saved to: {corrected_path}')
print(f'  Number of regions: {len(reg_names)}')
print(f'  Number of labeled vertices: {(vert_lab != 0).sum()}')
print(f'  Labels of the previously unlabeled vertices: {vert_lab[unlabeled_indices]}')
print('  Note: the values change between runs because the annotation is simulated.')
print('')

# Load the raw and the filled annotations onto the surface
surf = cltsurf.Surface(surf_file)
surf.load_annotation(cltfree.AnnotParcellation(parc_file=annot_file), parc_name=annot_id)
surf.load_annotation(cltfree.AnnotParcellation(parc_file=corrected_path), parc_name=f'{annot_id}_Filled')

# Visualize the raw and filled annotations side by side
print('Visualizing raw and filled annotations...')
plotter = cltvis.BrainPlotter()
plotter.plot(
    objs2plot=surf,
    views='medial',
    map_names=[annot_id, f'{annot_id}_Filled'],
)
```

### 1.7 `correct_vertexwise_parcellation`
<a id="annotparc-correct_parcellation"></a>

Correct parcellations by removing small clusters of vertices separated from the main cluster or labeled inside another region.

```python
######################## Testing correct_vertexwise_parcellation ###################################
import copy
import os
import tempfile

import numpy as np

import clabtoolkit.freesurfertools as cltfree
import clabtoolkit.surfacetools as cltsurf
import clabtoolkit.visualizationtools as cltvis

############## Loading example Annotation ###################################
freesurfer_home = os.environ.get('FREESURFER_HOME')
if freesurfer_home is None:
    # 1. Simulate a surface and save it to a temporary file
    surface = cltsurf.Surface.simulate_surface()
    surf_file = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.surface', delete=False).name
    surface.save_surface(filename=surf_file, overwrite=True)

    # 2. Simulate an annotation on the surface
    annot_parc = cltfree.AnnotParcellation.simulate_annotation(surface=surface, n_regions=5)

    # 3. Simulate a cortex label by excluding the last region
    codes = np.asarray(annot_parc.codes).ravel()
    last_code = np.unique(codes)[-1]
    labeled_indices = np.where(codes != last_code)[0]
    labeled_vertices = surface.mesh.points[labeled_indices]
    label_data = np.concatenate([labeled_indices[:, None], labeled_vertices], axis=1)

    # 4. Write the cortex label as a FreeSurfer .label file: [index, x, y, z, value]
    label_file = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.label', delete=False).name
    with open(label_file, 'w') as f:
        f.write('#!ascii label\n')
        f.write(f'{len(label_data)}\n')
        for idx, x, y, z in label_data:
            f.write(f'{int(idx)} {x:.6f} {y:.6f} {z:.6f} 0.500000\n')

    # 5. Introduce an isolated cluster labeled as source_code inside the region target_code
    source_code = np.unique(codes)[-2]
    target_code = np.unique(codes)[-3]
    target_indices = np.where(codes == target_code)[0]
    if len(target_indices) > 0:
        seed_vertex = target_indices[0]
        faces = surface.mesh.faces.reshape(-1, 4)[:, 1:]        # (n_faces, 3)
        mask = np.any(faces == seed_vertex, axis=1)            # faces containing the seed vertex
        isolated_vertices = np.unique(faces[mask])
        annot_parc.codes[isolated_vertices] = source_code
        print(f'Introduced an isolated cluster of {len(isolated_vertices)} vertices '
              f'(label {source_code}) inside region {target_code}.')

    # 6. Save the simulated annotation to a temporary file
    annot_file = tempfile.NamedTemporaryFile(prefix='lh.', suffix='.annot', delete=False).name
    annot_parc.save_annotation(annot_file)
    annot_id = 'Simulated'

else:
    subjects_dir = os.path.join(freesurfer_home, 'subjects')
    subject_id = 'fsaverage'
    hemi_id = 'lh'
    annot_id = 'aparc'
    surf_type = 'inflated'
    annot_file = os.path.join(subjects_dir, subject_id, 'label', f'{hemi_id}.{annot_id}.annot')
    label_file = os.path.join(subjects_dir, subject_id, 'label', f'{hemi_id}.cortex.label')
    surf_file = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.{surf_type}')

# Load the surface and the original annotation
surf = cltsurf.Surface(surf_file)
annot_parc = cltfree.AnnotParcellation(parc_file=annot_file)
surf.load_annotation(annot_parc, parc_name=annot_id)

# Test 1: Correct a copy of the annotation, so the original stays untouched
print('Test 1: Correcting the annotation (min_cluster_size=100)...')
annot_corrected = copy.deepcopy(annot_parc)
annot_corrected.correct_vertexwise_parcellation(surf_file, cortex_file=label_file, min_cluster_size=100)
n_changed = int((annot_corrected.codes != annot_parc.codes).sum())
print(f'  Vertices relabeled: {n_changed}')
print('')

surf.load_annotation(annot_corrected, parc_name=f'{annot_id}_Corrected')

# Visualize the original and corrected annotations side by side
print('Visualizing original and corrected annotations...')
plotter = cltvis.BrainPlotter()
plotter.plot(
    objs2plot=surf,
    views='medial',
    map_names=[annot_id, f'{annot_id}_Corrected'],
)
```

### 3.14 `get_version`
<a id="fstools-get_version"></a>

Get the FreeSurfer version number.

```python
######################## Testing get_version ###################################
import os
import re

import clabtoolkit.freesurfertools as cltfree

freesurfer_home = os.environ.get('FREESURFER_HOME')
if freesurfer_home is None:
    print("FreeSurfer installation not detected (FREESURFER_HOME is not set). Skipping this example.")
else:
    # Test 1: Get the local FreeSurfer version
    print('Test 1: Getting the local FreeSurfer version...')
    version = cltfree.get_version()
    print(f'  FreeSurfer version: {version}')
    print('')

    # Test 2: Compare versions as (major, minor, patch) tuples
    # (joining the digits, e.g. '7.10.0' -> 7100, gives wrong comparisons)
    print('Test 2: Comparing the version against 7.2.0...')
    if version:
        ver_tuple = tuple(int(v) for v in re.findall(r'\d+', str(version))[:3])
        print(f'  Version tuple : {ver_tuple}')
        print(f'  Is >= 7.2.0   : {ver_tuple >= (7, 2, 0)}')
    else:
        print('  Version could not be determined.')
    print('')
```
